```python
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python

# ===== Activation =====
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)
```

```python
# ===== Load data =====
data = fetch_california_housing()
X = data.data
y = data.target.reshape(-1,1)

# Normalize
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
# ===== Model =====
class NeuralNetwork:
    def __init__(self, input_size):
        self.w1 = np.random.randn(input_size,64) * 0.01
        self.w2 = np.random.randn(64,64) * 0.01
        self.w3 = np.random.randn(64,1) * 0.01

    def forward(self, X):
        self.z1 = np.dot(X, self.w1)
        self.a1 = relu(self.z1)

        self.z2 = np.dot(self.a1, self.w2)
        self.a2 = relu(self.z2)

        self.output = np.dot(self.a2, self.w3)
        return self.output

    def backward(self, X, y, lr=0.01):
        m = X.shape[0]

        delta3 = self.output - y
        dW3 = np.dot(self.a2.T, delta3) / m

        delta2 = np.dot(delta3, self.w3.T) * relu_derivative(self.a2)
        dW2 = np.dot(self.a1.T, delta2) / m

        delta1 = np.dot(delta2, self.w2.T) * relu_derivative(self.a1)
        dW1 = np.dot(X.T, delta1) / m

        self.w1 -= lr * dW1
        self.w2 -= lr * dW2
        self.w3 -= lr * dW3
```

```python
# ===== Train =====
nn = NeuralNetwork(X_train.shape[1])

for epoch in range(50):
    nn.forward(X_train)
    nn.backward(X_train, y_train, lr=0.01)

    loss = np.mean((y_train - nn.output)**2)
    if epoch % 10 == 0:
        print(f"Epoch {epoch}: Loss = {loss:.4f}")
```

```python
# ===== Evaluate =====
pred = nn.forward(X_test)

mse = np.mean((y_test - pred)**2)
rmse = np.sqrt(mse)

print("\nMSE:", mse)
print("RMSE:", rmse)
```
